*EXERCISE THREE (INVENTORY MANAGEMENT SYSTEM)*
=

```python
initial_stock = 250
sold_stock = 85

remaining_stock = initial_stock - sold_stock

reorder = ( remaining_stock <= 200)
print("Remaining stock: ", remaining_stock)
if reorder:
    print("Reorder Needed: ", reorder)
```

*EXERCISE FOUR (BANKING WITHDRAWAL CHECKER)*
=

```python
account_balance = 520000
amount_to_be_withdraw = 150000

remaining_balance = account_balance - amount_to_be_withdraw

approve_withdrawal = ( remaining_balance >= amount_to_be_withdraw)
print("Remaining Balance: ", remaining_balance)
if approve_withdrawal:
    print("Withdrawal Approved: ", approve_withdrawal)
    
```

**MINI PROJECT: SMART SHOPPING DISCOUNT CALCULATOR**
=

```python
#PURCHASES
laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000

total_purchase = laptop_bag + wireless_mouse + keyboard + usb_drive

discount_qualify = ( total_purchase >= 50000) 
print(total_purchase)
if discount_qualify:
    print("Discount Qualified: ", discount_qualify)
    
discount_amount = (10 / 100)* total_purchase
print("Discount Amount: ", discount_amount) 

final_amount = total_purchase - discount_amount
print("Final Amount: ", final_amount)

```
